```python
import pandas as pd
import os

comparison = 'B1_reintro_vs_B2_conA'  # Change for each comparison

df = pd.read_csv(f'{comparison}_significant.csv')

# Split by direction
upregulated = df[df['log2FoldChange'] > 0]['gene_name'].dropna()
downregulated = df[df['log2FoldChange'] < 0]['gene_name'].dropna()

# Save as TXT (one gene per line)
upregulated.to_csv(f'{comparison}_upregulated.txt', index=False, header=False)
downregulated.to_csv(f'{comparison}_downregulated.txt', index=False, header=False)
```
